```python
import pandas as pd
from pathlib import Path

RAW = Path("../data/raw")
df = pd.read_csv(RAW / "ICD-10.csv")
df.head()
```

```python
print(df.shape)
df.info()
df.isna().sum().sort_values(ascending=False)
```

```python
print("Valores de parent:")
print(df["parent"].dropna())

print("\nValores de Unnamed: 3:")
print(df["Unnamed: 3"].dropna())
```

```python
df = df.drop(columns=["parent", "Unnamed: 3"])
print(df.shape)
df.info
```

```python
df_filtrado = df[df.iloc[:, 0].astype(str).str.match(r'^[A-Za-z]\d{2}$')]
print(df_filtrado)
```

```python
valores_a_eliminar = ['id', 'text']
df = df[~df['id'].astype(str).str.lower().isin(valores_a_eliminar) & 
               ~df['text'].astype(str).str.lower().isin(valores_a_eliminar)]
print(df)
```

```python
df_primeros = df.groupby(df['id'].astype(str).str[0]).head(1)
print(df_primeros)
```

```python
df_fetus = df[df['text'].astype(str).str.contains('fetus', case=False, na=False)]
print(df_fetus)
```
